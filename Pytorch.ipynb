### **Autograd: Automatic Derivative calculator in Pytorch**
It is basically used for the purpose of performing the automatic differentiation for the backpropagation and other tensor operations. It enables gradient computation, which is essential for training machine learning models and for the optimization algorithms like gradient descent. For instance if we have max 10 neurons in the network we can calculate the gradients easily but when it comes to a large neural network then it becomes extremely difficult hence, autograd comes into picture for performing this computation. If we want we can even disable it while calculating the predictions.

```python
x = torch.tensor(5.0, requires_grad = True)
y = torch.tensor(10.0, requires_grad = True)
```

```python
def binary_cross_entropy(y_true, y_pred, eps=1e-15):
    y_pred = torch.clamp(y_pred, eps, 1 - eps)
    loss = -(y_true * torch.log(y_pred) +
             (1 - y_true) * torch.log(1 - y_pred)).mean()
    return loss
```

```python
class nnModel(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()
  def forward(self, features):
    out = self.linear(features)
    out = self.sigmoid(out)
    return out
  def loss_function(self, y_true, y_pred, eps=1e-7):
    y_pred = torch.clamp(y_pred, eps, 1 - eps)
    loss = -(y_true * torch.log(y_pred) +
             (1 - y_true) * torch.log(1 - y_pred)).mean()
    return loss
```

### **Dataset & Dataloader Class**

As the name suggests this class is responsible for the purpose of managing the data or the dataset that we will process before sending the data to the model for training and final results.

**1. Dataset is simply the collection of the object, which is responsible for the collection of the data samples and their corresponding lables, defining how to access individual items.**

**2. Dataloader on the other hand is the utility that wraps the dataset to provide an efficient, iterable interface for accessing data in batches, with options for shuffling and parallel loading.**

**The Role of a Dataset**

The Dataset class (specifically torch.utils.data.Dataset in PyTorch) serves as a blueprint for your data. It decouples the data access logic from the model
training logic. To create a custom dataset, you typically override two methods:

1. __len__(self): Returns the total number of samples in the dataset.
2. __getitem__(self, idx): Loads and returns a single data sample and its label at the specified index. This is where data transformations (e.g., resizing images, tokenizing text) are often applied to individual samples.

**The Role of a DataLoader**

The DataLoader (torch.utils.data.DataLoader) is the workhorse for feeding data to your model's training loop. It handles the complexities of efficient data loading and offers several key features:

**1. Automatic Batching:** It groups individual samples fetched from the Dataset into mini-batches, which is crucial for efficient model training on GPUs.

**2. Shuffling:** For training data, the DataLoader can shuffle the data indices at the beginning of each epoch to prevent the model from memorizing the order of samples, leading to better generalization.

**3. Parallel Data Loading:** The num_workers parameter enables multiprocessing, allowing several subprocesses to load batches concurrently and significantly speeding up the data pipeline, especially with heavy I/O operations.

**4. Memory Pinning:** If a GPU is available, the pin_memory=True argument can be used to copy tensors into "pinned" memory, which accelerates the transfer of data from the CPU to the GPU.

**5. Custom Collation:** A collate_fn argument allows for custom logic to combine samples into a batch, useful for handling variable-length data like text sequences that require padding.

```python
class mynn(nn.Module):
  def __init__(self, input_dim, output_dim, num_hidden_layers, neurons_per_layers, dropout_rate):
    super().__init__()
    layers = []
    for i in range(num_hidden_layers):
      layers.append(nn.Linear(input_dim, neurons_per_layers))
      layers.append(nn.BatchNorm1d(neurons_per_layers))
      layers.append(nn.ReLU())
      layers.append(nn.Dropout(dropout_rate))
      input_dim = neurons_per_layers
    layers.append(nn.Linear(neurons_per_layers, output_dim))
    self.model = nn.Sequential(*layers)
  def forward(self, x):
    return self.model(x)
```

```python
# Objective function
def objective(trial):
  # next hyperperameter values from the search space
  num_hidden_layers = trial.suggest_int("num_hidden_layers", 1, 5)
  neurons_per_layer = trial.suggest_int("neurons_per_layer", 8, 32, step = 8)
  epochs = trial.suggest_int("epochs", 10, 50, step = 10)
  learning_rate = trial.suggest_float("learning_rate", 1e-5, 1e-1, log = True)
  dropout_rate = trial.suggest_float("dropout_rate", 0.2, 0.5, step = 0.1)
  batch_size = trial.suggest_categorical("batch_size", [32, 64, 128])
  optimizer_name = trial.suggest_categorical("optimizer", ["Adam", "SGD", "RMSprop"])
  activation_function = trial.suggest_categorical("activation_function", ["ReLU", "Tanh", "Sigmoid"])
  weight_decay = trial.suggest_float("weight_decay", 1e-5, 1e-3, log = True)


  # dataset init

  train_loader = DataLoader(train_dataset, batch_size = batch_size, shuffle = True, pin_memory=True)
  test_loader = DataLoader(test_dataset, batch_size = batch_size, shuffle = False, pin_memory=True)

  # model init
  input_dim = 784
  output_dim = 10

  model = mynn(input_dim, output_dim, num_hidden_layers, neurons_per_layer, dropout_rate)
  model.to(device)
  # params init
  learning_rate = 0.01
  epochs = 50

  # optimizer selection
  criteria = nn.CrossEntropyLoss()
  # optimizer = optim.SGD(model.parameters(), lr = learning_rate, weight_decay = weight_decay)

  if optimizer_name == "Adam":
    optimizer = optim.Adam(model.parameters(), lr = learning_rate, weight_decay = weight_decay)
  elif optimizer_name == "SGD":
    optimizer = optim.SGD(model.parameters(), lr = learning_rate, weight_decay = weight_decay)
  else:
    optimizer = optim.RMSprop(model.parameters(), lr = learning_rate, weight_decay = weight_decay)


  # training loop
  for epoch in range(epochs):
    total_epoch_loss = 0
    for batch_features, batch_labels in train_loader:
      batch_features, batch_labels = batch_features.to(device), batch_labels.to(device)
      y_pred = model(batch_features)
      loss = criteria(y_pred, batch_labels)
      optimizer.zero_grad()
      loss.backward()
      optimizer.step()
      total_epoch_loss += loss.item()
    total_epoch_loss /= len(train_loader)
  print(f"Epoch {epoch + 1}, Loss: {total_epoch_loss}")

  # evaluation
  model.eval()
  total = 0
  correct = 0
  with torch.no_grad():
    for batch_features, batch_labels in test_loader:
      batch_features, batch_labels = batch_features.to(device), batch_labels.to(device)
      y_pred = model(batch_features)
      _, predicted = torch.max(y_pred, 1)
      total += batch_labels.size(0) # shape[0] will also work the same.
      correct += (predicted == batch_labels).sum().item()
    accuracy = correct / total
  return accuracy
```

### **Convolutional Neural Network**

```python
class CustomDataset(Dataset):
  def __init__(self, X, y):
    self.X = torch.tensor(X, dtype = torch.float32).reshape(-1, 1, 28, 28)
    self.y = torch.tensor(y, dtype = torch.long)
  def __len__(self):
    return len(self.X)
  def __getitem__(self, idx):
    return self.X[idx], self.y[idx]
```

```python
train_Data = CustomDataset(X_train, y_train)
test_Data = CustomDataset(X_test, y_test)
train_Dataloader = DataLoader(train_Data, batch_size = 128, shuffle = True, pin_memory = True)
test_Dataloader = DataLoader(test_Data, batch_size = 128, shuffle = False, pin_memory = True)
```

```python
class CNNModel(nn.Module):
  def __init__(self, input_features):
    super().__init__()
    self.features = nn.Sequential(
        nn.Conv2d(input_features, 32, kernel_size = 3, stride = 1, padding = 'same'),
        nn.ReLU(),
        nn.BatchNorm2d(32),
        nn.MaxPool2d(kernel_size = 2, stride = 2),

        nn.Conv2d(32, 64, kernel_size = 3, padding = 'same'),
        nn.ReLU(),
        nn.BatchNorm2d(64),
        nn.MaxPool2d(kernel_size = 2, stride = 2),
    )
    self.classifier = nn.Sequential(
        nn.Flatten(),
        nn.Linear(64 * 7 * 7, 128),
        nn.ReLU(),
        nn.Dropout(0.4),
        nn.Linear(128, 64),
        nn.ReLU(),
        nn.Dropout(0.4),
        nn.Linear(64, 10)
    )

  def forward(self, x):
    x = self.features(x)
    x = self.classifier(x)

    return x
```

```python
learning = 0.01
epochs = 100
```

```python
model = CNNModel(1)
model.to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.AdamW(model.parameters(), lr = learning, weight_decay = 0.01)
```

```python
for epoch in range(epochs):
    total_epoch_loss = 0
    for batch_features, batch_labels in train_Dataloader:
        batch_features, batch_labels = batch_features.to(device), batch_labels.to(device)
        y_pred = model(batch_features)
        loss = criterion(y_pred, batch_labels)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total_epoch_loss += loss.item()
    total_epoch_loss /= len(train_Dataloader)
    print(f"Epoch {epoch + 1}, Loss: {total_epoch_loss}")
```

### **Transfer Learning**
It is a concept where in, a model is first trained on a large dataset, when a good accuracy is achieved, this model is saved and finally, it can be used for other different but related tasks. Instead of training a model from scratch we can easily use this pretrained for related tasks, this saved both time and cost, also increases the efficiency of the model

**Working of Transfer Learning**
1. Pretraining on a Large Dataset.
2. Fine Tuning on new task.

```python
from PIL import Image
class CustomisedDataset(Dataset):
  def __init__(self, features, labels, transform):
    self.features = features
    self.labels = labels
    self.transform = transform
  def __len__(self):
    return len(self.features)
  def __getitem__(self, index):
    image = self.features[index].reshape(28,28)
    image = image.astype(np.uint8)
    image = np.stack([image] * 3, axis = -1)
    # Explicitly specify mode='RGB' to ensure it's treated as a 3-channel image
    image = Image.fromarray(image, mode='RGB')
    image = self.transform(image)
    return image, torch.tensor(self.labels[index], dtype = torch.long)
```
